```python
def get_cohort_dataframe(condition, start, end, defects = False, daughters_leaving = 0, df = df):
    return df[
        (df["Condition"] == condition) &
        (df["Frame_onset_hrs"] >= start) &
        (df["Frame_hrs"] < end) &
        (df["any_true"] == defects) &
        (df["N_leaving_offspring"] == daughters_leaving) &
        (df["Generation"] == 1)
    ]
```

```python
def get_chi2_dataframe(df):
    counts = (
        df.groupby(["Dataset", "Metaphase_arrested"])["Number_of_daughters_in_mitosis"]
        .value_counts()
        .rename("count")
        .reset_index()
    ).rename(columns = {
        "Dataset": "replicate",
        "Metaphase_arrested": "time_bin",
        "Number_of_daughters_in_mitosis": "category"
    })
    return counts
```

```python
def plot_daughter_fate(df, plotheight = 200, plotwidth = 190):
    Bars = (
        alt.Chart(df, height = plotheight, width = plotwidth)
        .mark_bar(size = 23)
        .encode(
            x = alt.X("Metaphase_arrested:N", title = ""),
            y = alt.Y("count()", stack = "normalize", axis = alt.Axis(format = '%'), title = "% of mother cells in cohort"),
            color = alt.Color("Number_of_daughters_in_mitosis:O", title = "N proliferating daughters", scale = alt.Scale(scheme = "darkred"))
        )
    )
    return Bars
```

```python
# Get counts
def get_counts(df, name):

    counts = (
        df.groupby("Metaphase_arrested")["Number_of_daughters_in_mitosis"]
          .value_counts()
          .rename("count")
          .reset_index()
    )
    
    # Compute total per group
    counts["group_total"] = counts.groupby("Metaphase_arrested")["count"].transform("sum")
    
    # Compute percentage within each group
    counts["percentage"] = counts["count"] / counts["group_total"] * 100

    counts.to_csv(root + f"Statistics/Daughter_Fates_{name}.csv")
    
    return counts



def getstatistics(counts, name):
    # Because data are counts in a contingency table, the most statistically appropriate and stable method for pairwise bin comparisons is
    # Chi-square test on the 3×2 table (categories × bins)
    # This avoids the multinomial regression instability.

    table = counts.pivot_table(
    index = "category",
    columns = "time_bin",
    values = "count",
    aggfunc = "sum",
    fill_value = 0
    )
    expanded = counts.loc[counts.index.repeat(counts["count"])].drop(columns = "count")
    global_chi2, global_p, global_dof, globale_expected = chi2_contingency(table)
    
    def compare_bins(counts_df, bin1, bin2):
        subset = counts_df[counts_df["time_bin"].isin([bin1, bin2])]
        table = subset.pivot_table(
            index = "category",
            columns = "time_bin",
            values = "count",
            aggfunc = "sum",
            fill_value = 0
        )

        chi2, p, dof, expected = chi2_contingency(table)

        return p

    bins = counts["time_bin"].unique()

    results = []

    for b1, b2 in combinations(bins, 2):
        p = compare_bins(counts, b1, b2)
        results.append((b1, b2, p))
    
    pairwise = pd.DataFrame(results, columns = ["bin1","bin2","p"])
    pairwise["p_adj_Benjamini-Hochberg"] = multipletests(pairwise["p"], method = "fdr_bh")[1] # this is the Multiple testing correction (Benjamini–Hochberg (FDR))
    pairwise["Global chi2 p value"] = global_p
    print(f"Chi2 + FDR BH Pairwise results for {name} were saved.")
    pairwise.to_csv(root + f"Statistics/Chi2_pvalues_Fates_{name}.csv")
```

```python
def Scatter_bin(dataframe, x, y, color, x_title, y_title, binextent, binstep,
            Circlesize = 18, 
            CircleOpacity = 0.2,  
            Scatter_width = 300, 
            Scatter_height = 200
               ):
    colourscheme = "viridis"    
    Scatter_binned = alt.Chart(
        data = dataframe, 
        width = Scatter_width, 
        height = Scatter_height
    ).mark_circle(
        opacity = 1,
        size = 125
    ).encode(
        alt.X(x, title = x_title, bin = alt.Bin(extent = binextent, step = binstep)),
        alt.Y(f'median({y})', title = y_title, bin = False),
        color = alt.Color(
            color, scale = alt.Scale(scheme = colourscheme)
        ) 
    )
    
    Error_Scatterbinned = alt.Chart(
            data = dataframe
    ).mark_errorbar(extent = "iqr").encode(
        alt.X(x, title = x_title, bin = alt.Bin(extent = binextent, step = binstep)),
        alt.Y(f'mean({y})', title = y_title, bin = False),
        color = alt.Color(
            color, scale = alt.Scale(scheme = colourscheme)
        ) 
    ) 
    SCATTERBIN = Error_Scatterbinned + Scatter_binned
    return SCATTERBIN
```

```python
# Error analysis

def function(x):
    if x["Mitotic Death"] == True or x["Mitotic Death"] == "True":
        return "Mitotic death"
    elif x["any_true"] == True:
        return "Mitotic defect / errors"
    else:
        return "Normal division"
        
def plot_Individual_Daughters_error(df, category, barwidth, y_max, plotwidth = 450, plotheight = 250):
    bar = alt.Chart(df).mark_bar(width = barwidth).encode(
        x = alt.X(
            "Unique_Source_ID:N",
            title = "Individual divisions",
            sort = alt.EncodingSortField(
                field = "Mitotic_duration_mins",
                order = "ascending"
            ),
            axis = alt.Axis(
                labels = False, 
                ticks = False
            )
        ),
        y = alt.Y("Mitotic_duration_mins:Q", title = "Mitotic duration (min)", scale = alt.Scale(domain = [0, y_max])), 
        color = alt.Color(category, title = f"{category}?", scale = alt.Scale(scheme = "cividis")),
    )

    chart = bar.properties(
        width = plotwidth,
        height = plotheight,
    )
    return chart


df_ctrl_all["Function"] = df_ctrl_all.apply(function, axis = 1)
df_augmin_all["Function"] = df_augmin_all.apply(function, axis = 1)
```

```python
# Mitotic fate / Error percentage bar charts (Figure 6C') - BASED ON MEAN OF THE DATASETS

df_6Clinha = pd.concat([df_augmin_all, df_ctrl_all])

# Count occurrences
counts = (
    df_6Clinha.groupby(["Dataset", "Condition", "Augmin_depletion_bin", "Function"])
      .size()
      .reset_index(name = "count")
)

# Convert to percentages within each Dataset/Condition/Bin
counts["percentage"] = (
    counts.groupby(["Dataset", "Condition", "Augmin_depletion_bin"])["count"]
          .transform(lambda x: x / x.sum() * 100)
)

summary = (
    counts.groupby(["Condition", "Augmin_depletion_bin", "Function"])
          .agg(
              mean_percentage=("percentage", "mean"),
              sem = ("percentage", lambda x: x.std(ddof=1) / np.sqrt(len(x)))
          )
          .reset_index()
)

summary["mean_percentage_norm"] = (
    summary.groupby(["Condition", "Augmin_depletion_bin"])["mean_percentage"]
           .transform(lambda x: x / x.sum() * 100)
)

#summary.to_csv(root + "/Statistics/Figure6Clinha.csv")

# Create plots 
base = alt.Chart(summary, height = 60, width = 250)
bars = base.mark_bar(size = 17).encode(
    y=alt.Y("Condition:N", title="Condition"),
    x=alt.X("mean_percentage_norm:Q", title="Mean Percentage", scale = alt.Scale(domain = [0, 100])),
    color="Function:N"
)

error_bars = base.mark_errorbar().encode(
    y="Condition:N",
    x=alt.X("mean_percentage_norm:Q", title="Mean Percentage", scale = alt.Scale(domain = [0, 100])),
    xError="sem:Q",
    color="Function:N"
)

chart = alt.layer(
    bars,
    error_bars
).facet(
    column="Augmin_depletion_bin:N"
)

#chart
```

```python
cols_no_death = [
    "Laggings", 
    "Massive Missegregation", 
    "DNA bridge", 
    "Misaligned", 
    "Multipolar Division", 
    'Cytokinesis Failure',
    'Slippage', 
   # 'Mitotic Death'
       ]


df_short = df_6Clinha[cols_no_death + ['Dataset', 'Condition', 'Metaphase_arrested', 'Augmin_depletion_bin', "Frame_hrs", "N_leaving_offspring"]]
df_short.to_csv(root + "Statistics/Figure6Cii_SourceData.csv")

def plot_binned_errors(timebin, df = df_short):

    #grouped_bin1 = df_short[df_short['Augmin_depletion_bin'] == timebin].groupby(['Dataset', 'Condition'])[cols_no_death].mean().mul(100)
    grouped_bin1 = df_short[df_short['Augmin_depletion_bin'] == timebin].groupby(['Condition'])[cols_no_death].mean().mul(100)
    
    mean_df_bin1 = grouped_bin1.groupby('Condition').mean()
    #sem_df_bin1 = grouped_bin1.groupby('Condition').sem().fillna(0)
    
    plot_df = (
        mean_df_bin1.reset_index().melt(id_vars = 'Condition', var_name = 'Category', value_name = 'Mean')
    #    .merge(
    #        sem_df_bin1.reset_index().melt(id_vars = 'Condition', var_name = 'Category', value_name = 'SEM'),
    #        on = ['Condition', 'Category']
    #    )
    )
    
    plot_df["Mean"] = plot_df.Mean.astype(float)
    
    # Stacking order
    order = [c for c in mean_df.columns if c != 'Normal'] + ['Normal']
    plot_df['Category'] = pd.Categorical(plot_df['Category'], categories = order, ordered = True)
    
    # sort by condition then the category order so cumsum matches visual order
    plot_df['order_index'] = plot_df['Category'].cat.codes
    plot_df = plot_df.sort_values(['Condition', 'order_index']).reset_index(drop = True)
    
    # compute bottoms & tops (explicit stacking)
    plot_df['bottom'] = plot_df.groupby('Condition')['Mean'].cumsum() - plot_df['Mean']
    plot_df['top']    = plot_df['bottom'] + plot_df['Mean']
    
    # compute error bar extents relative to each slice
   # plot_df['err_low']  = plot_df['bottom'] + (plot_df['Mean'] - plot_df['SEM'])
   # plot_df['err_high'] = plot_df['bottom'] + (plot_df['Mean'] + plot_df['SEM'])
    
    bars = alt.Chart(plot_df).mark_bar(size = 17).encode(
        y = alt.Y('Condition:N', title = 'Condition'),
        x = alt.X('bottom:Q', title = '% of divisions', scale = alt.Scale(domain = [0, 35])),
        x2 = 'top:Q',
        color = alt.Color('Category:N', sort = order, legend = alt.Legend(title = 'Category')),
        tooltip = [
            'Condition:N',
            'Category:N',
            alt.Tooltip('Mean:Q', format = '.2f'),
            alt.Tooltip('SEM:Q', format = '.2f')
        ]
    ).properties(width = 250, height = 60)
    
 #  err = alt.Chart(plot_df).mark_rule(strokeWidth = 2).encode(
 #      y = 'Condition:N',
 #      x = 'err_low:Q',
 #      x2 = 'err_high:Q',
 #      color = alt.Color('Category:N', sort = order, legend = None),
 #      tooltip = [
 #          'Condition:N',
 #          'Category:N',
 #          alt.Tooltip('Mean:Q', format = '.2f'),
 #          alt.Tooltip('SEM:Q', format = '.2f')
 #      ]
 #  )
    
    chart = bars# + err
    
    return chart
```

```python
def plot_Individual_Daughters(df, barwidth = 1.8, plotwidth = 600, plotheight = 250):
    bar = alt.Chart(df).mark_bar(width = barwidth).encode(
        x = alt.X(
            "Unique_Source_ID:N",
            title = "Individual daughter cells",
            sort = alt.EncodingSortField(
                field = "Mother_Mitotic_duration_mins",
                order = "ascending"
            ),
            axis = alt.Axis(
                labels = False, 
                ticks = False
            )
        ),
        y = alt.Y("Mother_Mitotic_duration_mins:Q", title = "Mother mitotic duration (min)", scale = alt.Scale(domain = [0, 700])),
        color = alt.Color("Cell_Cycle_hours:Q", title = "Daughter cell cycle duration (hours)", scale = alt.Scale(scheme = "blueorange")),
    )

    chart = bar.properties(
        width = plotwidth,
        height = plotheight,
    )

    return chart
```
